```python
class Stack:
    """
    고정된 크기(10)를 가진 스택 클래스
    """
    def __init__(self):
        self.stack = []
        self.limit = 10  # 스택의 최대 크기 제한

    def push(self, item):
        """
        스택에 데이터를 추가하는 함수
        데이터가 10개가 차면 추가되지 않고 경고 메시지 출력
        """
        if len(self.stack) >= self.limit:
            print(f"Warning: 스택이 가득 찼습니다. '{item}'을(를) 추가할 수 없습니다.")
        else:
            self.stack.append(item)
            print(f"Push: {item}")

    def pop(self):
        """
        가장 마지막에 추가된 내용을 가져오는 함수
        내용이 없으면 경고 메시지 출력
        """
        if self.empty():
            print('Warning: 스택이 비어있어 가져올 데이터가 없습니다.')
            return None
        else:
            item = self.stack.pop()
            return item

    def peek(self):
        """
        마지막 내용을 삭제하지 않고 확인하는 함수
        """
        if self.empty():
            print('스택이 비어있습니다.')
            return None
        return self.stack[-1]

    def empty(self):
        """
        전체 내용이 비었는지 확인하는 함수 (True/False 반환)
        """
        return len(self.stack) == 0

    def visualize(self):
        """
        [보너스 과제] 스택의 현재 상태를 직관적으로(시각화) 보여주는 함수
        텍스트 기반의 그래픽(ASCII Art)으로 표현
        """
        print('\n=== Current Stack State ===')
        if self.empty():
            print('|     (Empty)     |')
        else:
            # 위에서부터 아래로 출력 (역순 순회)
            for item in reversed(self.stack):
                print(f'| {str(item).center(15)} |')
        print('-----------------------')
        print(f'Total: {len(self.stack)}/{self.limit}\n')


# --- 실행 및 테스트 코드 ---
if __name__ == '__main__':
    # 1. 스택 생성
    my_stack = Stack()

    # 2. Push 테스트 (1부터 12까지 넣어본다 -> 10개 제한 확인)
    print('--- 데이터 입력 테스트 (Push) ---')
    for i in range(1, 13):
        item_name = f'Data-{i}'
        my_stack.push(item_name)

    # 3. 시각화 확인 (보너스 과제)
    my_stack.visualize()

    # 4. Peek 테스트
    print(f'Peek (맨 위 데이터 확인): {my_stack.peek()}')

    # 5. Pop 테스트 (LIFO 확인)
    print('\n--- 데이터 꺼내기 테스트 (Pop) ---')
    for _ in range(3):
        pop_item = my_stack.pop()
        print(f'Pop: {pop_item}')

    # 6. 최종 상태 시각화
    my_stack.visualize()
    
    # 7. 모두 비우기 테스트 (Underflow 확인)
    print('--- 모두 비우기 ---')
    while not my_stack.empty():
        my_stack.pop()
        
    # 비어있는 상태에서 Pop 시도
    my_stack.pop()
```
