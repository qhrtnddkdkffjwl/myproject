```python
import cv2
import os

def task1_basic():
    print("=== [과제 1] OpenCV 기초 실행 ===")
    
    # 1. 경로 설정 (OPENCV 폴더 내부)
    base_dir = 'OPENCV'
    img_path = os.path.join(base_dir, 'image.jpg')
    video_path = os.path.join(base_dir, 'video.mp4')

    # --- 이미지 출력 ---
    if os.path.exists(img_path):
        img = cv2.imread(img_path)
        cv2.imshow('Image Viewer', img)
        print(f"이미지 출력: {img_path} (아무 키나 누르면 닫힘)")
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print(f"오류: 이미지를 찾을 수 없습니다 ({img_path})")

    # --- 동영상 출력 ---
    if os.path.exists(video_path):
        cap = cv2.VideoCapture(video_path)
        print(f"동영상 재생: {video_path} (종료: ESC)")
        
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            
            cv2.imshow('Video Player', frame)
            if cv2.waitKey(33) == 27: # ESC
                break
        
        cap.release()
        cv2.destroyAllWindows()
    else:
        print(f"오류: 동영상을 찾을 수 없습니다 ({video_path})")

if __name__ == "__main__":
    task1_basic()
```

```python
import cv2
import datetime
import os

def task2_control():
    print("=== [과제 2] 단축키 제어 시작 ===")
    print("[ESC]:종료  [Ctrl+Z]:캡처  [Ctrl+X]:녹화시작  [Ctrl+C]:녹화중지")

    # 경로 설정
    base_dir = 'OPENCV'
    video_path = os.path.join(base_dir, 'video.mp4')
    
    if not os.path.exists(video_path):
        print(f"오류: 파일을 찾을 수 없습니다 ({video_path})")
        return

    cap = cv2.VideoCapture(video_path)
    is_recording = False
    recorder = None
    fourcc = cv2.VideoWriter_fourcc(*'mp4v') # 코덱

    while True:
        ret, frame = cap.read()
        if not ret:
            cap.set(cv2.CAP_PROP_POS_FRAMES, 0) # 무한 반복
            continue

        # 녹화 상태 표시
        if is_recording:
            cv2.circle(frame, (30, 30), 10, (0, 0, 255), -1) # 빨간 점
            if recorder: recorder.write(frame)

        cv2.imshow('Control Player', frame)

        key = cv2.waitKey(33) & 0xFF

        # [ESC] 종료
        if key == 27:
            break
        
        # [Ctrl+Z] 캡처
        elif key == 26:
            now = datetime.datetime.now().strftime("%Y%m%d_%H-%M-%S")
            save_path = os.path.join(base_dir, f"capture_{now}.jpg")
            cv2.imwrite(save_path, frame)
            print(f"📸 캡처 저장: {save_path}")

        # [Ctrl+X] 녹화 시작
        elif key == 24:
            if not is_recording:
                is_recording = True
                now = datetime.datetime.now().strftime("%Y%m%d_%H-%M-%S")
                save_path = os.path.join(base_dir, f"record_{now}.mp4")
                
                h, w = frame.shape[:2]
                fps = cap.get(cv2.CAP_PROP_FPS) or 30
                recorder = cv2.VideoWriter(save_path, fourcc, fps, (w, h))
                print(f"🔴 녹화 시작: {save_path}")

        # [Ctrl+C] 녹화 중지
        elif key == 3:
            if is_recording:
                is_recording = False
                if recorder:
                    recorder.release()
                    recorder = None
                print("⏹ 녹화 완료")

    if recorder: recorder.release()
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    task2_control()
```
