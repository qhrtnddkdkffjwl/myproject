```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    # [수행과제] 추가 함수 (맨 앞, 중간, 맨 뒤 가능)
    def insert(self, data, position=-1):
        new_node = Node(data)
        
        # 1. 맨 앞에 추가하거나 리스트가 비어있을 때
        if position == 0 or self.head is None:
            new_node.next = self.head
            self.head = new_node
            return

        # 2. 중간 혹은 맨 뒤에 추가
        current = self.head
        curr_pos = 0
        
        # 원하는 위치의 바로 앞 노드(prev)까지 이동
        # position이 -1이면 맨 끝으로 이동
        while current.next:
            if position != -1 and curr_pos == position - 1:
                break
            current = current.next
            curr_pos += 1
            
        # 노드 연결 (삽입)
        new_node.next = current.next
        current.next = new_node

    # [수행과제] 삭제 함수
    def delete(self, data):
        if self.head is None:
            print("리스트가 비어있습니다.")
            return

        # 헤드 노드를 삭제하는 경우
        if self.head.data == data:
            self.head = self.head.next
            return

        # 중간 혹은 끝 노드 삭제
        current = self.head
        while current.next:
            if current.next.data == data:
                current.next = current.next.next # 연결 건너뛰기
                return
            current = current.next
        
        print(f"'{data}'를 찾을 수 없습니다.")

    # [보너스 과제] 순차적으로 가져오는 함수
    def get_list(self):
        result = []
        current = self.head
        while current:
            result.append(current.data)
            current = current.next
        return result

# --- 테스트 코드 (음악 목록) ---
if __name__ == "__main__":
    print("=== 단순 연결 리스트 (Playlist) ===")
    mylist = LinkedList()
    
    # 음악 추가
    mylist.insert("아이유 - 밤편지")        # 맨 앞 (처음)
    mylist.insert("NewJeans - Hype Boy")    # 맨 뒤 (기본)
    mylist.insert("BTS - Dynamite", 1)      # 중간 (인덱스 1)
    mylist.insert("aespa - Next Level", 0)  # 맨 앞 지정
    
    print("현재 리스트:", mylist.get_list())
    
    # 음악 삭제
    print("\n'BTS - Dynamite' 삭제 중...")
    mylist.delete("BTS - Dynamite")
    
    print("최종 리스트:", mylist.get_list())
```

```python
import time
import os

# 실제 음악 재생을 위해 필요 (없으면 설치: pip install pygame)
# 파일이 없으면 시뮬레이션 모드로 작동하도록 예외처리 함
try:
    import pygame
    pygame.mixer.init()
    HAS_PYGAME = True
except ImportError:
    HAS_PYGAME = False
    print("pygame이 설치되지 않아 소리 재생은 스킵하고 로그만 출력합니다.")

class CircularNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.file_path = None # 보너스 과제용 파일 경로

class CircularList:
    def __init__(self):
        self.head = None
        self.current_node = None # 현재 재생 위치 포인터

    # [수행과제] 추가 함수 (임의의 위치 - 여기서는 Append 방식 사용)
    def insert(self, data, file_path=None):
        new_node = CircularNode(data)
        new_node.file_path = file_path # 보너스 과제용
        
        if self.head is None:
            self.head = new_node
            new_node.next = self.head # 자기 자신을 가리킴 (원형)
            self.current_node = self.head
        else:
            # 마지막 노드를 찾아서 새 노드 연결
            temp = self.head
            while temp.next != self.head:
                temp = temp.next
            temp.next = new_node
            new_node.next = self.head # 새 노드가 다시 헤드를 가리킴

    # [수행과제] 삭제 함수
    def delete(self, data):
        if self.head is None: return

        curr = self.head
        prev = None
        
        while True:
            if curr.data == data:
                # 1. 헤드를 삭제하는 경우
                if curr == self.head:
                    # 노드가 하나뿐일 때
                    if curr.next == self.head:
                        self.head = None
                        self.current_node = None
                        return
                    
                    # 마지막 노드를 찾아 헤드 변경 사항 반영
                    last = self.head
                    while last.next != self.head:
                        last = last.next
                    self.head = curr.next
                    last.next = self.head
                
                # 2. 중간/끝 노드 삭제
                else:
                    prev.next = curr.next
                return
            
            prev = curr
            curr = curr.next
            if curr == self.head: # 한 바퀴 돌았으면 종료
                break
        print(f"'{data}'를 찾을 수 없습니다.")

    # [수행과제] 다음 항목 가져오기 (플레이어의 Next 버튼)
    def get_next(self):
        if self.current_node:
            data = self.current_node.data
            file = self.current_node.file_path
            
            # 포인터 이동
            self.current_node = self.current_node.next
            return data, file
        return None, None

    # [수행과제] 검색 함수
    def search(self, data):
        if self.head is None: return False
        
        curr = self.head
        while True:
            if curr.data == data:
                return True
            curr = curr.next
            if curr == self.head:
                break
        return False

    # [보너스 과제] 음악 플레이어 실행
    def play_music(self, count=5):
        print("\n=== 🎵 음악 플레이어 시작 (5곡 재생) ===")
        
        for i in range(count):
            title, path = self.get_next()
            if title:
                print(f"▶️ 재생 중 [{i+1}/{count}]: {title}")
                
                # 실제 파일 재생 로직 (파일이 존재할 경우)
                if HAS_PYGAME and path and os.path.exists(path):
                    try:
                        pygame.mixer.music.load(path)
                        pygame.mixer.music.play()
                        # 음악 길이만큼 대기하거나 3초 미리듣기
                        time.sleep(3) 
                        pygame.mixer.music.stop()
                    except Exception as e:
                        print(f"   (파일 재생 오류: {e})")
                else:
                    # 파일이 없으면 시뮬레이션
                    time.sleep(1) 
            else:
                print("재생할 목록이 없습니다.")
                break

# --- 테스트 코드 ---
if __name__ == "__main__":
    player = CircularList()
    
    # 1. 음악 등록 (제목, 파일경로)
    # 실제 mp3 파일이 있다면 경로를 넣어보세요. (예: "C:/Music/song.mp3")
    player.insert("Queen - Bohemian Rhapsody", "queen.mp3")
    player.insert("Beetles - Let it be", "letitbe.mp3")
    player.insert("Michael Jackson - Beat It", "beatit.mp3")
    
    # 2. 검색 테스트
    print(f"검색 'Queen - Bohemian Rhapsody': {player.search('Queen - Bohemian Rhapsody')}")
    print(f"검색 'K-Pop': {player.search('K-Pop')}")
    
    # 3. [보너스] 플레이어 작동 (원형 구조라 계속 순환됨)
    player.play_music(count=5)
```
