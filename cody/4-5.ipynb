```python
class Node:
    """
    이진 트리의 노드를 구성하는 클래스
    """
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    """
    이진 탐색 트리(BST) 클래스 (보너스 과제 포함)
    """
    def __init__(self):
        self.root = None

    def insert(self, data):
        """
        새로운 원소를 추가하는 함수
        """
        if self.root is None:
            self.root = Node(data)
        else:
            self._insert_recursive(self.root, data)

    def _insert_recursive(self, current_node, data):
        """
        재귀적으로 위치를 찾아 삽입하는 내부 도우미 함수
        """
        if data < current_node.data:
            if current_node.left is None:
                current_node.left = Node(data)
            else:
                self._insert_recursive(current_node.left, data)
        else:
            if current_node.right is None:
                current_node.right = Node(data)
            else:
                self._insert_recursive(current_node.right, data)

    def find(self, data):
        """
        원하는 값의 존재 유무를 확인하는 탐색 함수
        """
        return self._find_recursive(self.root, data)

    def _find_recursive(self, current_node, data):
        if current_node is None:
            return False
        if data == current_node.data:
            return True
        elif data < current_node.data:
            return self._find_recursive(current_node.left, data)
        else:
            return self._find_recursive(current_node.right, data)

    def delete(self, data):
        """
        특정 원소를 삭제하는 함수
        """
        self.root = self._delete_recursive(self.root, data)

    def _delete_recursive(self, current_node, data):
        if current_node is None:
            return current_node

        # 삭제할 데이터를 찾아 이동
        if data < current_node.data:
            current_node.left = self._delete_recursive(current_node.left, data)
        elif data > current_node.data:
            current_node.right = self._delete_recursive(current_node.right, data)
        else:
            # 삭제할 노드를 찾은 경우 (Case 1, 2, 3)

            # Case 1 & 2: 자식이 없거나 하나만 있는 경우
            if current_node.left is None:
                return current_node.right
            elif current_node.right is None:
                return current_node.left

            # Case 3: 자식이 둘 다 있는 경우
            # 오른쪽 서브트리에서 가장 작은 값을 찾아 현재 노드를 대체
            min_val = self._find_min_value(current_node.right)
            current_node.data = min_val
            # 대체한 값을 가진 원래 노드를 삭제
            current_node.right = self._delete_recursive(current_node.right, min_val)

        return current_node

    def _find_min_value(self, node):
        """
        서브트리에서 가장 작은 값을 찾는 함수 (삭제 시 후계자 탐색용)
        """
        current = node
        while current.left is not None:
            current = current.left
        return current.data

    def print_tree(self):
        """
        트리 내용을 순서대로 출력 (In-order Traversal) - 테스트용
        """
        print('Tree Elements:', end=' ')
        self._inorder_traversal(self.root)
        print()

    def _inorder_traversal(self, node):
        if node:
            self._inorder_traversal(node.left)
            print(node.data, end=' ')
            self._inorder_traversal(node.right)

# --- 실행 및 테스트 코드 ---
if __name__ == '__main__':
    # 1. 이진 탐색 트리 생성 (이름: binarytree)
    binarytree = BinarySearchTree()

    # 2. 데이터 삽입 테스트 (insert)
    print('=== 데이터 삽입 ===')
    input_data = [50, 30, 20, 40, 70, 60, 80]
    for n in input_data:
        binarytree.insert(n)
    
    binarytree.print_tree()

    # 3. 데이터 탐색 테스트 (find)
    print('\n=== 데이터 탐색 ===')
    targets = [40, 90]
    for t in targets:
        found = binarytree.find(t)
        print(f'{t} 존재 여부: {found}')

    # 4. 데이터 삭제 테스트 (delete)
    print('\n=== 데이터 삭제 ===')
    
    # Case 1: 자식이 없는 노드 삭제 (20)
    print('Deleting 20 (Leaf Node)...')
    binarytree.delete(20)
    binarytree.print_tree()

    # Case 2: 자식이 하나인 노드 삭제 (30) - 30은 40을 자식으로 가짐
    print('Deleting 30 (Node with 1 child)...')
    binarytree.delete(30)
    binarytree.print_tree()

    # Case 3: 자식이 둘인 노드 삭제 (50) - Root 삭제
    print('Deleting 50 (Node with 2 children)...')
    binarytree.delete(50)
    binarytree.print_tree()
```
